## Unificar Dataframe de Customers, Services e Contracts

```python
# Analise de quantidade de registro de casa dataframe
len(df_customers) 
```

```python
len(df_contracts)
```

```python
len(df_services)
```

```python
# Renomeando colunas
df_services.rename(columns={'customerID': 'IDCliente'}, inplace=True)
```

```python
df_customers.info()
```

```python
df_services.info()
```

```python
# Unificar Dataframe de Customers com services, criando um terceiro Dataframe
df_temp = df_customers.merge(df_services, on=['IDCliente'], how="inner", validate="many_to_many")
```

```python
df_temp.info()
```

```python
df_temp.head(5)
```

```python
# Unificar df_temp com Contracts, usando colunas de junção com nomes distintos
df_churn_temp = df_temp.merge(df_contracts, left_on=['IDCliente'], right_on=['customerID'], how="inner", validate="many_to_many")
```

```python
df_churn_temp.info()
```

```python
df_contracts.rename(columns={'customerID': 'IDCliente'}, inplace=True)
```

```python
df_churn_temp = df_temp.merge(df_contracts, on=['IDCliente'], how="inner", validate="many_to_many")
```

```python
df_churn_temp.info()
```

```python
df_churn = df_customers.merge(df_services, on=['IDCliente'], how="inner", validate="many_to_many").merge(df_contracts, on=['IDCliente'], how="inner", validate="many_to_many")
```
